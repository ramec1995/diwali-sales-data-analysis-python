```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Diwali Sales Data.csv', encoding='unicode_escape')
df
```

```python
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df
```

```python
df.info()
```

```python
df['Amount'] = df['Amount'].astype('int')
df.info()
```

```python
df.rename(columns={'Marital_Status':'Married'}, inplace=True)
df
```

```python
df[['Age', 'Orders', 'Amount']].describe()
```

```python
sns.countplot(x='Gender', data=df, hue='Gender')
for bar in plt.gca().containers:
    plt.gca().bar_label(bar)
plt.show()
```

```python
Gender_sales = df.groupby('Gender')['Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(x='Gender', y='Amount', data=Gender_sales, hue='Gender')
plt.show()
```

From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men.

```python
sns.countplot(x='Age Group', data=df, hue='Gender')
for bars in plt.gca().containers:
    plt.gca().bar_label(bars)
plt.show()
```

```python
age_group_sales = df.groupby('Age Group')['Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(x='Age Group', y='Amount', data=age_group_sales)
plt.show()
```

From above graphs we can see that most of the buyers are of age group between 26-35 yrs female.

```python
statewise_orders = df.groupby('State')['Orders'].sum().sort_values(ascending=False).reset_index().head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x='State', y='Orders', data=statewise_orders, hue='State')
plt.xticks(rotation=90)
plt.show()
```

```python
statewise_sales = df.groupby('State')['Amount'].sum().sort_values(ascending=False).reset_index().head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x='State', y='Amount', data=statewise_sales, hue='State')
plt.xticks(rotation=90)
plt.show()
```

From above graphs we can see that most of the orders & total sales/amount are from Uttar Pradesh, Maharashtra and Karnataka respectively.

```python
sns.countplot(x='Married', data=df)
for bars in plt.gca().containers:
    plt.gca().bar_label(bars)
plt.show()
```

```python
Marital_sales = df.groupby(['Married','Gender'])['Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(x='Married', y='Amount', data=Marital_sales, hue='Gender')
plt.show()
```

From above graphs we can see that most of the buyers are single (women) and they have high purchasing power.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Occupation', data=df, hue='Occupation')
for bars in plt.gca().containers:
    plt.gca().bar_label(bars)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
Occupation_sales = df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(x='Occupation', y='Amount', data=Occupation_sales, hue='Occupation')
plt.xticks(rotation=90)
plt.show()
```

From above graphs we can see that most of the buyers are working in IT, Healthcare and Aviation sector.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Product_Category', data=df, hue='Product_Category')
for bars in plt.gca().containers:
    plt.gca().bar_label(bars)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
category_sales = df.groupby('Product_Category')['Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(x='Product_Category', y='Amount', data=category_sales, hue='Product_Category')
plt.xticks(rotation=90)
plt.show()
```

From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category.





```python
sns.countplot(x='Zone', data=df, hue='Zone')
for bars in plt.gca().containers:
    plt.gca().bar_label(bars)
plt.show()
```

```python
Zone_sales = df.groupby('Zone')['Amount'].sum().sort_values(ascending=False).reset_index()
sns.barplot(x='Zone', y='Amount', data=Zone_sales, hue='Zone')
plt.show()
```

From above graphs we can see that most of the products were sold in Central Zone followed by Southern and Western respectively.

```python
Product_sales = df.groupby('Product_ID')['Orders'].sum().sort_values(ascending=False).reset_index().head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x='Product_ID', y='Orders', data=Product_sales, hue='Product_ID')
plt.xticks(rotation=90)
plt.show()
```

From above graphs we can see that the most performing product is P00265242.

#Conclusion:
Married women age group 26-35 yrs from UP, Maharastra and Karnataka working in IT, Healthcare and Aviation are more likely to buy products from Food, Clothing and Electronics category
